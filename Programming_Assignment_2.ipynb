  A. REPRODUCIBLE NORMALIZATION PROBLEM

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
X
```

```python
x = np.mean(X)
x
```

```python
s = np.std(X)
s
```

```python
Z = (X-x)/s
Z
```

```python
X_Normalized = Z
print(X_Normalized)
```

```python
x_Normalized = np.mean(X_Normalized)
print(x_Normalized)
```

```python
s_Normalized = np.std(X_Normalized)
print(s_Normalized)
```

```python
np.save("X_Normalized.npy", X_Normalized)
```

B. CUBES DIVISIBLE BY 4 PROBLEM

```python
c = np.arange (1,101,1)
c
```

```python
c.shape
```

```python
c = np.power(c, 3)
c
```

```python
 c.reshape (10,10)
```

```python
div_by_4 = c[c % 4 == 0]
print(div_by_4)
```

```python
div_by_4.shape
```

```python
np.save("div_by_4.npy",div_by_4)
```

C. ABOVE-MEAN SQUARES PROBLEM

```python
S = np.arange (1,37).reshape(6,6)
S

```

```python
S = np.power(S,2)
S
```

```python
S_mean = np.mean(S)
S_mean
```

```python
above_mean = S[S_mean < S]
above_mean
```

```python
above_mean.shape
```

```python
np.save("above_mean.npy",above_mean)
```
